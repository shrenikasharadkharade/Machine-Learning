```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("/content/Country-data.csv")
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
print("Duplicate Values")
df.duplicated().sum()
```

```python
print("First 5 records: ")
print(df.head())
```

```python
x = df.drop('country', axis=1)
x
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
print(x_scaled[:5])
```

```python
x_scaled = pd.DataFrame(
scaler.fit_transform(x),
columns = x.columns
)
print(x_scaled.head())
```

```python
KMeans = KMeans(n_clusters= 3, random_state=42, n_init=10)

```

```python
KMeans.fit(x_scaled)
```

```python
df['Cluster'] = KMeans.labels_
print("Countries with Cluster: ")
print(df[['country','Cluster']].head(15))
```

```python
print("Number of countries in each cluster: ")
print(df['Cluster'].value_counts().sort_index())
```

```python
new_sample = pd.DataFrame(
[[10,40,7,35,30000,5,75,2,15000]],
columns= x.columns
)
new_sample_scaled = pd.DataFrame(
scaler.transform(new_sample),
columns= x.columns
)
new_cluster = KMeans.predict(new_sample_scaled)
```

```python
print(" New country sample: ")
print(new_sample)
print(" New Country belongs to cluster: ",new_cluster[0])
```
